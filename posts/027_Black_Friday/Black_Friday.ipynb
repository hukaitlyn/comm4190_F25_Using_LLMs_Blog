## **Introduction:**

Black Friday always feels a little chaotic and a little tempting at the same time. There are so many stores claiming to have the best deals that it becomes hard to tell what is actually worth buying. This year I wondered whether an LLM could cut through the noise and guide me toward smarter choices. Instead of scrolling endlessly or running around from store to store I decided to treat AI like a personal shopping assistant and see what it could actually do. Could an LLM help me find the right stores plan my route and figure out which deals were actually real bargains?


## **Prompt:**

> I want you to act as my Black Friday shopping assistant with a focus on fashion and beauty products. Please compare deals from stores like Zara Aritzia Ulta Sephora and any other stores with strong Black Friday sales. Highlight real discounts and flag anything that is misleading or not worth it. Help me decide which stores are best for clothing basics trends jackets shoes beauty sets fragrances and tools. If I want to shop in person give me a simple suggested order of stores based on efficiency.

## **Response:**

Absolutely! I’ll be your Black Friday shopping assistant focused on fashion and beauty products. I’ve gathered the most current and realistic Black Friday deals from major retailers including Zara, Aritzia, Ulta, Sephora, and other notable stores known for strong sales. I’ll break everything down by category—clothing basics, trends, jackets, shoes, beauty sets, fragrances, and beauty tools—then give you an in-person shopping strategy if you plan to go store to store.

Let’s get you the best value without falling for marketing tricks. 🔍

🛍️ Black Friday 2023-2024 Highlights (Fashion & Beauty)

FASHION: CLOTHING + SHOES

Zara  
Deal Overview: Historically underwhelming. Black Friday discount is usually 30% off select items online and in-store—limited selection and often from slower-moving styles.  
Categories:
- Basics: Limited selection; not ideal.
- Trends: Okay selection if lucky, but sells out quickly.
- Jackets: Not many discounted; size availability limited.
- Shoes: Limited sizes and discounted items.

🏷️ Verdict: Worth browsing if you’re specifically into Zara aesthetics, but don’t expect deep discounts or broad selection. Shop early on Black Friday if you go.

Aritzia  
Deal Overview: One of the stronger Black Friday fashion events. Usually 10%-50% off most everything sitewide, tiers vary by brand/category.  
Categories:
- Basics: 10–20% off on staples like Babaton & Wilfred—reasonable.
- Trends: Good markdowns on seasonal pieces and styles.
- Jackets: Coats get the best discounts (20–50%). Super Puff and wool coats included but go fast.
- Shoes/Boots: Limited, but discounts here too.

🏷️ Verdict: A must for in-person shopping if you're after quality basics or winterwear. Markdowns are real (even for popular pieces), but inventory vanishes fast.

Other Stores to Consider:
- Abercrombie & Fitch: 25-30% off sitewide. Great for basics, denim, outerwear.
- Mango: Usually 30% off everything with code. Comparable to Zara but often better selection and pricing.
- Nordstrom: Price Matching + steep discounts on designer basics, seasonal outerwear, and shoes (especially boots and sneakers). Excellent value.

BEAUTY: PRODUCTS + FRAGRANCES

Ulta Beauty  
Deal Overview: Impressive category-wide deals, especially on tools and gift sets. Big brands are participating.  
Categories:
- Beauty Sets: 30–50% off name-brand sets—great value.
- Tools: 40% off hot tools (T3, Revlon, Drybar); some early deals already live.
- Fragrance: Limited, but some 25–35% off sets including Ariana Grande, Philosophy, etc.
- Skincare/Makeup: NYX, Tarte, Urban Decay, IT Cosmetics—heavily discounted.

🏷️ Verdict: Excellent spot for mass and prestige hybrid shopping. Their doorbusters are real—shop on the earlier side or online for inventory.

Sephora  
Deal Overview: More selective—usually 25% off select brands, smaller bundles, and some 50% off daily deals.  
Categories:
- Beauty Sets: Good, but many are Sephora collection or curated minis—evaluate cost-per-use.
- Fragrance: Less impressive—10–20% off more common (brands like Nest, Replica Minis).
- Tools: Some markdowns, but often better deals at Ulta or Amazon.
- Makeup/Skincare: Premium brands like Tatcha, Urban Decay, and KVD sometimes 30–50% off for 1 day only.

🏷️ Verdict: Best for beauty insiders with early access. Steeper day-specific deals, but inventory and selection won't match Ulta for tools or gift sets.

Other Notables:
- Dermstore: 30% off skincare brands like Sunday Riley, EltaMD, Paula’s Choice.
- Amazon: Top-notch for beauty tools—Revlon One-Step, PMD, ghd all usually 30–50% off.
- Target: Very strong beauty tool + mass skincare deals (the same Revlon dryers/tools).

⚠️ Deals that are often Misleading:
- Sephora Fragrance Samplers: Can be fun, but more “try before you commit” than cost-saving.
- Zara “select item” markdowns: Not the pieces that sell well or trend-forward.
- Clothing shops that advertise “up to 70% off” but only include out-of-season or low-selling inventory.

💎 Best Stores by Category

Clothing Basics — Aritzia (quality per dollar), Abercrombie, Mango  
Trends — Zara (if you dig hard), Mango, H&M (budget quick wins)  
Jackets — Aritzia (seriously worth it this time), Abercrombie, Nordstrom  
Shoes — Nordstrom Sale section, DSW (select boots and sneakers marked down heavily), Steve Madden (50% sitewide)  
Beauty Sets — Ulta hands down, Sephora for brand loyalists  
Fragrance — Nordstrom/Ulta combo (gift sets + value sizes), not Sephora  
Beauty Tools — Ulta, Amazon, Dermstore  

🛒 In-Person Shopping Strategy (Suggested Route)

Start early (8–9 AM ideally) — wear something comfy!

1. Aritzia → inventory sells FAST, especially The Super Puff and basic layers. Lines will grow.
2. Zara → shorter haul and overlapping mall location — browse for trend-specific pieces.
3. Nordstrom → for outerwear/boots/fragrance/beauty crossovers.
4. Ulta → better inventory midday, solid staff help, more beauty doorbusters.
5. Sephora → limited specials left unless you’re after specific brand deals.

Optional: End at Target or Amazon (online) for tools, backups, and any missed items—especially Dyson/T3/PMD tools.

🧠 Pro Tips:
- Pre-add desired items to wishlists on websites.
- Follow major retailers’ apps (Zara restricts early access through app).
- Check if stores offer buy online/pickup in store to save time.
- Use a budget tracker quickly via app (like Mint) to avoid impulse purchases stacked up.

⏳ Ideal Shopping Split (if Part Online, Part In Person):
Friday morning: shop Aritzia and Ulta in-store  
Friday lunch: mobile scroll & purchase top tools/fragrance/beauty kits online  
Saturday: Zara restocks occasionally — recheck app

Let me know your specific wishlist, preferences, or planned locations for even more efficient personalized guidance. Happy deal hunting! 🛍💄👢

## **Reflection:**

When I first ran the prompt I was honestly impressed by how much information the LLM gave me. It broke everything down by category explained each store’s usual Black Friday patterns and even gave me a shopping route which was more detailed than I expected. But at the same time I felt a little disappointed because it only focused on the specific stores I listed. I was hoping it would go beyond my examples and give me more of a master list with everything I could possibly shop from. In my head I kind of imagined a giant spreadsheet of every sale across fashion and beauty with links categories and side by side comparisons. Instead the model stayed pretty close to the stores I mentioned and filled in just a few extras. It was still helpful and organized but not as all encompassing as I was expecting which made me realize I might need to prompt it differently next time or be more explicit about wanting that big picture overview.
